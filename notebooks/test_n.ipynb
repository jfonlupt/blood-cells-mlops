```python
import pandas as pd
import ast
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. Charger ton DataFrame
# ============================================================

# Exemple :
df = pd.read_csv('/home/remi/jul26_bds_blood_cells/results/leaderboard.csv')

# ============================================================
# 2. Garder uniquement les expériences à 8 epochs
# ============================================================

df_n = df[df["train.nb_epoch"] >= 50].copy()

import matplotlib.pyplot as plt
from pathlib import Path

# Meilleur modèle selon le F1 validation
best_row = df_n.loc[df_n["test_f1_macro"].idxmax()]
display(df_n.sort_values(by="test_f1_macro", ascending=False).head(5))
# Récupérer le code du run
best_code = Path(best_row["run_dir"]).name

# Chemin de la matrice de confusion
cm_path = Path("../results") / best_code / "figures" / "metrics_test_confusion_matrix.png"

print("Meilleur modèle :", best_code)
print("test_f1_macro   :", best_row["test_f1_macro"])
print("Image           :", cm_path)

# Affichage
img = plt.imread(cm_path)

plt.figure(figsize=(10, 10))
plt.imshow(img)
plt.axis("off")
plt.title(f"Matrice de confusion — F1 test = {best_row['test_f1_macro']:.3f}")
plt.show()

cm_path = Path("../results") / best_code / "figures" / "training_loss.png"
img = plt.imread(cm_path)
plt.figure(figsize=(10, 10))
plt.imshow(img)
plt.axis("off")
plt.show()

# ============================================================
# 3. Extraire les paramètres de l'architecture
# ============================================================

def parse_blocks(x):
    """
    Convertit :
    "[{'depth': 2, 'channels': 32, 'kernel': 3}, ...]"
    en liste Python.
    """
    if isinstance(x, list):
        return x

    try:
        return ast.literal_eval(x)
    except Exception:
        return []


df_n["blocks"] = df_n["model.architecture.blocks"].apply(parse_blocks)


# Nombre de blocks
df_n["n_blocks"] = df_n["blocks"].apply(len)


# Profondeur des blocks
# Ici on prend la profondeur moyenne.
df_n["block_depth"] = df_n["blocks"].apply(
    lambda x: sum(b["depth"] for b in x) / len(x)
    if len(x) > 0 else None
)


# Nombre de channels
# On prend le dernier nombre de channels.
df_n["channels"] = df_n["blocks"].apply(
    lambda x: x[-1]["channels"]
    if len(x) > 0 else None
)


# Kernel
# On prend le kernel du premier block.
df_n["kernel"] = df_n["blocks"].apply(
    lambda x: x[0]["kernel"]
    if len(x) > 0 else None
)


# ============================================================
# 4. Nombre de couches du classifier
# ============================================================

def parse_classifier(x):
    if isinstance(x, list):
        return x

    try:
        return ast.literal_eval(x)
    except Exception:
        return []


df_n["classifier"] = df_n["model.architecture.classifier"].apply(
    parse_classifier
)


# Nombre de couches neuronales
df_n["n_dense_layers"] = df_n["classifier"].apply(len)


# Nombre de neurones
# Pour plusieurs couches :
# [128, 128, 128] → 128,128,128
df_n["neurons_per_layer"] = df_n["classifier"].apply(
    lambda x: "-".join(map(str, x))
)


df_n["use_batchnorm"] = df_n["model.architecture.use_batchnorm"].astype(bool)

df_n["use_gap"] = df_n["model.architecture.use_gap"].astype(bool)

# ============================================================
# 5. Fonction pour tracer l'impact d'un paramètre
# ============================================================

def scatter_impact(df, parameters, metric="val_f1_macro"):

    n = len(parameters)

    fig, axes = plt.subplots(
        1,
        n,
        figsize=(5 * n, 5),
        squeeze=False
    )

    axes = axes[0]

    for ax, parameter in zip(axes, parameters):

        # Tous les essais
        sns.scatterplot(
            data=df,
            x=parameter,
            y=metric,
            s=100,
            ax=ax
        )

        # Moyenne par valeur
        mean = (
            df.groupby(parameter, observed=True)[metric]
            .mean()
            .reset_index()
            .sort_values(parameter)
        )

        sns.lineplot(
            data=mean,
            x=parameter,
            y=metric,
            marker="o",
            linewidth=2,
            ax=ax
        )

        ax.set_xlabel(parameter)
        ax.set_ylabel(metric)
        ax.set_title(parameter)
        ax.grid(True, alpha=0.3)

    fig.suptitle(
        "Impact des paramètres d'architecture — 5 epochs",
        fontsize=16
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 6. Graphiques
# ============================================================
parameters = [
    "n_blocks",
    "block_depth",
    "kernel",
    "channels",
    "n_dense_layers",
    "neurons_per_layer",
    "use_batchnorm",
    "use_gap",
]

scatter_impact(
    df_n,
    parameters
)
```
