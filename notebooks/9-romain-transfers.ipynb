## Comparaison des configurations

`grid_search_transfer_2` croise 5 backbones (ResNet-50, EfficientNet-B0, DenseNet-121,
ConvNeXt-Tiny, ViT-Base/16) avec 6 sous-expériences : linear probing, fine-tuning complet
(sweep de `backbone_lr_mult`), gel partiel, contrôle poids aléatoires, normalisation
(ImageNet vs dataset) et architecture de la tête. On les sépare ici en groupes pour
pouvoir comparer les configurations entre elles.

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

%config InlineBackend.figure_format = "retina"
sns.set_theme(style="ticks")

BACKBONE_ORDER = [
    "resnet50",
    "efficientnet_b0",
    "densenet121",
    "convnext_tiny",
    "vit_base_patch16_224.augreg2_in21k_ft_in1k",
]
BACKBONE_LABEL = {
    "resnet50": "ResNet-50",
    "efficientnet_b0": "EfficientNet-B0",
    "densenet121": "DenseNet-121",
    "convnext_tiny": "ConvNeXt-Tiny",
    "vit_base_patch16_224.augreg2_in21k_ft_in1k": "ViT-Base/16",
}
BACKBONE_LABEL_ORDER = [BACKBONE_LABEL[b] for b in BACKBONE_ORDER]
BACKBONE_COLORS = dict(
    zip(BACKBONE_LABEL_ORDER, sns.color_palette("colorblind", len(BACKBONE_ORDER)))
)


def assign_group(r):
    """Retrouve la sous-expérience (bloc de configs/grids/architecture/transfer.yaml) d'une run."""
    if r["model.transfer.freeze"] == "all" and r["model.transfer.head_hidden"] == "[]":
        return "G1_linear_probe"
    if (
        r["model.transfer.freeze"] == "none"
        and r["model.transfer.pretrained"]
        and r["transform.norm_stats"] == "imagenet"
    ):
        return "G2_finetune_full"
    if r["model.transfer.freeze"] in ("1", "2"):
        return "G3_partial_freeze"
    if not r["model.transfer.pretrained"]:
        return "G4_random_init"
    if r["transform.norm_stats"] == "dataset":
        return "G5_norm_stats"
    if r["model.transfer.freeze"] == "all" and r["model.transfer.head_hidden"] != "[]":
        return "G6_head_arch"
    return "OTHER"


df_tr["group"] = df_tr.apply(assign_group, axis=1)
df_tr["backbone"] = df_tr["model.transfer.backbone"].map(BACKBONE_LABEL)
assert (df_tr["group"] != "OTHER").all(), (
    "certaines runs ne correspondent à aucun bloc de la grille"
)
df_tr["group"].value_counts()
```

```python
from pathlib import Path as _Path

ROOT = _Path.cwd().parent if _Path.cwd().name == "notebooks" else _Path.cwd()
FIG_DIR = ROOT / "reports" / "figures" / "report"
FIG_DIR.mkdir(parents=True, exist_ok=True)
DPI = 200


def save_fig(fig, name: str) -> None:
    """Sauvegarde dans reports/figures/report/ et laisse la figure s'afficher."""
    out = FIG_DIR / name
    fig.savefig(out, dpi=DPI, bbox_inches="tight", facecolor="white")
    print(f"  → {out.relative_to(ROOT)}")
```

### 1. Backbones en linear probing (backbone gelé)

Compare les 5 backbones tête nue, tous figés, sans dégel : isole la qualité des features
pré-entraînées, indépendamment de tout fine-tuning.

```python
METRIC_LABEL = {
    "val_f1_macro": "F1 macro",
    "val_mcc": "MCC",
    "val_min_critical_recall": "Rappel critique min.",
}

g1 = df_tr[df_tr.group == "G1_linear_probe"]
g1_long = g1.melt(
    id_vars="backbone", value_vars=list(METRIC_LABEL), var_name="metric", value_name="score"
)
g1_long["metric"] = g1_long["metric"].map(METRIC_LABEL)

fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(
    data=g1_long,
    x="backbone",
    y="score",
    hue="metric",
    order=BACKBONE_LABEL_ORDER,
    hue_order=list(METRIC_LABEL.values()),
    palette="colorblind",
    ax=ax,
)
ax.set_xlabel("")
ax.set_ylabel("Score (validation)")
ax.set_title("Linear probing (backbone gelé) — comparaison des backbones", fontweight="bold")
ax.tick_params(axis="x", rotation=15)
for label in ax.get_xticklabels():
    label.set_ha("right")
ax.set_ylim(0, 1.12)
ax.legend(frameon=False, title="", loc="upper left", ncol=3)
sns.despine(ax=ax)
fig.tight_layout()
```

### 2. Fine-tuning complet : effet du LR discriminatif du backbone

Backbone entièrement dégelé (`freeze=none`) : effet de `backbone_lr_mult` (LR backbone
= LR tête × mult) sur le score moyen (F1 macro / MCC / rappel critique).

```python
g2 = df_tr[df_tr.group == "G2_finetune_full"]
fig, ax = plt.subplots(figsize=(8, 5.5))
sns.lineplot(
    data=g2,
    x="model.transfer.backbone_lr_mult",
    y="val_avg",
    hue="backbone",
    hue_order=BACKBONE_LABEL_ORDER,
    palette=BACKBONE_COLORS,
    marker="o",
    ax=ax,
)
ax.set_xscale("log")
ax.set_xlabel("Multiplicateur de LR du backbone (échelle log)")
ax.set_ylabel("Score moyen (F1 / MCC / rappel critique)")
ax.set_title("Fine-tuning complet — effet du LR discriminatif du backbone", fontweight="bold")
ax.legend(frameon=False, fontsize=9, title="")
sns.despine(ax=ax)
fig.tight_layout()
```

### 3. Stratégie de gel : linear probing vs gel partiel vs fine-tuning complet

Pour chaque backbone, meilleure configuration (parmi les `backbone_lr_mult` testés)
selon la stratégie de gel : entièrement gelé, dégel des 1 ou 2 derniers blocs, ou
entièrement dégelé.

```python
strategy_rows = []
for b in BACKBONE_ORDER:
    lbl = BACKBONE_LABEL[b]
    sub_all = df_tr[(df_tr["model.transfer.backbone"] == b) & (df_tr.group == "G1_linear_probe")]
    strategy_rows.append(
        {"backbone": lbl, "strategy": "Gelé (linear probing)", "val_avg": sub_all["val_avg"].max()}
    )
    for fz, label in [("1", "Partiel (1 dernier bloc)"), ("2", "Partiel (2 derniers blocs)")]:
        sub = df_tr[
            (df_tr["model.transfer.backbone"] == b) & (df_tr["model.transfer.freeze"] == fz)
        ]
        strategy_rows.append({"backbone": lbl, "strategy": label, "val_avg": sub["val_avg"].max()})
    sub_none = df_tr[(df_tr["model.transfer.backbone"] == b) & (df_tr.group == "G2_finetune_full")]
    strategy_rows.append(
        {
            "backbone": lbl,
            "strategy": "Dégelé (fine-tuning complet)",
            "val_avg": sub_none["val_avg"].max(),
        }
    )
strat_df = pd.DataFrame(strategy_rows)
strategy_order = [
    "Gelé (linear probing)",
    "Partiel (1 dernier bloc)",
    "Partiel (2 derniers blocs)",
    "Dégelé (fine-tuning complet)",
]

fig, ax = plt.subplots(figsize=(11, 5.5))
sns.barplot(
    data=strat_df,
    x="backbone",
    y="val_avg",
    hue="strategy",
    order=BACKBONE_LABEL_ORDER,
    hue_order=strategy_order,
    palette="colorblind",
    ax=ax,
)
ax.set_xlabel("")
ax.set_ylabel("Meilleur score moyen (val)")
ax.set_title("Stratégie de gel — meilleure configuration par backbone", fontweight="bold")
ax.tick_params(axis="x", rotation=15)
for label in ax.get_xticklabels():
    label.set_ha("right")
ax.set_ylim(0, 1.14)
ax.legend(
    frameon=False, fontsize=8.5, ncol=4, title="", loc="upper center", bbox_to_anchor=(0.5, 1.0)
)
sns.despine(ax=ax)
fig.tight_layout()
save_fig(fig, "transfer_grid_freeze_strategy.png")
```

### 4. Apport du pré-entraînement

Contrôle : mêmes architectures, poids ImageNet vs poids aléatoires, à `freeze=none`
et `backbone_lr_mult=0.1` (config de référence).

## Runs finaux

Les meilleurs backbones identifiés par `grid_search_transfer_2` (EfficientNet-B0,
DenseNet-121, ViT-Base/16), chacun avec sa meilleure configuration (fine-tuning complet,
`backbone_lr_mult` et optimiseur propres à chaque backbone, tête linéaire), sont
ré-entraînés une fois sur le jeu complet et évalués sur le test
(`run_name = final_{backbone}_type`). ConvNeXt-Tiny, tout aussi compétitif sur la grille
mais initialement non retenu, a depuis fait l'objet d'un run final complémentaire.

```python
import json

FINAL_RUNS = {
    "efficientnet_b0": "826fe85f",
    "densenet121": "abed1581",
    "vit_base_patch16_224.augreg2_in21k_ft_in1k": "59390a83",
    "convnext_tiny": "9c861225",
}
RUNS_DIR = ROOT / "results" / "runs"

final_hist, final_test, final_per_class = {}, {}, {}
for backbone, run_hash in FINAL_RUNS.items():
    run_dir = RUNS_DIR / run_hash
    hist = json.loads((run_dir / "history.json").read_text())
    final_hist[backbone] = hist[0] if isinstance(hist, list) else hist
    final_test[backbone] = json.loads((run_dir / "metrics_test.json").read_text())
    final_per_class[backbone] = pd.read_csv(run_dir / "figures" / "metrics_test_per_class.csv")

final_summary = pd.DataFrame(
    {
        BACKBONE_LABEL[b]: {
            "n_epochs": len(final_hist[b]["train_loss"]),
            "test_acc": final_test[b]["test_acc"],
            "test_f1_macro": final_test[b]["test_f1_macro"],
            "test_mcc": final_test[b]["test_mcc"],
            "test_top3_acc": final_test[b]["test_top3_acc"],
            "test_critical_screening_ap": final_test[b]["test_critical_screening_ap"],
            "test_grouped_critical_recall": final_test[b]["test_grouped_critical_recall"],
            "test_min_critical_recall (Promyelocyte)": final_per_class[b]
            .set_index("class")
            .loc["Promyelocyte", "recall"],
        }
        for b in FINAL_RUNS
    }
).T
final_summary
```

```python
fig, axes = plt.subplots(2, len(FINAL_RUNS), figsize=(4 * len(FINAL_RUNS), 7), sharex="col")
for col, b in enumerate(FINAL_RUNS):
    h = final_hist[b]
    epochs = range(1, len(h["train_loss"]) + 1)
    lbl = BACKBONE_LABEL[b]

    ax = axes[0, col]
    ax.plot(epochs, h["train_loss"], label="train", color=BACKBONE_COLORS[lbl])
    ax.plot(epochs, h["val_loss"], label="val", color=BACKBONE_COLORS[lbl], linestyle="--")
    ax.set_title(lbl, fontweight="bold")
    ax.set_ylabel("Perte" if col == 0 else "")
    ax.legend(frameon=False, fontsize=8)

    ax2 = axes[1, col]
    ax2.plot(epochs, h["train_f1"], label="train", color=BACKBONE_COLORS[lbl])
    ax2.plot(epochs, h["val_f1"], label="val", color=BACKBONE_COLORS[lbl], linestyle="--")
    ax2.set_xlabel("Époque")
    ax2.set_ylabel("F1 macro" if col == 0 else "")

for ax in axes.flat:
    sns.despine(ax=ax)
fig.suptitle(
    "Courbes d'entraînement des runs finaux — écart train/val croissant, arrêt anticipé bien avant le budget de 100 époques",
    fontweight="bold",
    y=1.02,
)
fig.tight_layout()
save_fig(fig, "transfer_final_training_curves.png")
```

```python
pc_rows = []
for b in FINAL_RUNS:
    sub = final_per_class[b][["class", "recall"]].copy()
    sub["backbone"] = BACKBONE_LABEL[b]
    pc_rows.append(sub)
pc_df = pd.concat(pc_rows, ignore_index=True)
class_order = final_per_class["densenet121"].sort_values("recall")["class"].tolist()

fig, ax = plt.subplots(figsize=(10, 5.5))
sns.barplot(
    data=pc_df,
    x="class",
    y="recall",
    hue="backbone",
    order=class_order,
    hue_order=[BACKBONE_LABEL[b] for b in FINAL_RUNS],
    palette="colorblind",
    ax=ax,
)
ax.axhline(0.95, color="#999999", linestyle=":", lw=1)
ax.set_xlabel("")
ax.set_ylabel("Rappel (test)")
ax.set_title("Rappel de test par classe — runs finaux", fontweight="bold")
ax.tick_params(axis="x", rotation=25)
for label in ax.get_xticklabels():
    label.set_ha("right")
ax.legend(
    frameon=False,
    fontsize=8.5,
    ncol=len(FINAL_RUNS),
    title="",
    loc="upper center",
    bbox_to_anchor=(0.5, 1.0),
)
sns.despine(ax=ax)
fig.tight_layout()
save_fig(fig, "transfer_final_per_class_recall.png")
```

## Comparaison élargie : grille existante + sweep complémentaire

Rémi a mené en parallèle son propre sweep de transfert (`results/leaderboard_remi.csv`,
run `transfer-sweep`), qui couvre les mêmes sous-expériences (linear probing, fine-tuning
complet, gel partiel, poids aléatoires, normalisation) mais ajoute **2 backbones absents
de `grid_search_transfer_2`** : EfficientNetV2-S et Swin-Tiny.

On refait ici les mêmes comparaisons que ci-dessus, sur l'union des deux sources, avec
deux réserves méthodologiques importantes :

- **Protocole non strictement identique** : le sweep de Rémi tourne majoritairement en
  `batch_size=124` (voire 256/512 pour quelques runs sur le jeu complet), contre
  `batch_size=32` dans `grid_search_transfer_2`, à budget d'époques égal (10). Sur les
  4 backbones testés dans les deux sweeps, les scores de Rémi sont systématiquement
  2 à 4 points plus bas au même `backbone_lr_mult` — plausiblement un effet du batch
  (moins de pas de gradient à budget d'époques fixe) plutôt qu'une vraie différence
  d'architecture. Les comparaisons **entre les deux sources** doivent donc être lues
  avec prudence ; les comparaisons **à l'intérieur d'une même source** restent fiables.
- **Rappel critique indisponible** : `val_min_critical_recall` n'est renseigné que sur
  les 10 runs les plus récents du sweep de Rémi (10/53). Le score moyen utilisé plus haut
  (`val_avg`, moyenne F1 macro / MCC / rappel critique) mélangerait donc des moyennes à 2
  et 3 termes selon les lignes. On utilise ici `val_avg2 = moyenne(F1 macro, MCC)`,
  disponible pour toutes les runs des deux sources.

Les 3 runs non comparables du sweep de Rémi (2 runs sur le jeu complet à 50 époques, 1
run de 1 époque) sont exclus pour ne garder que le protocole `15 000 images / 10 époques`
commun aux deux sources.

```python
remi_path = ROOT / "results" / "leaderboard_remi.csv"
df_remi = pd.read_csv(remi_path, parse_dates=["timestamp"])

df["leaderboard_source"] = "leaderboard"
df_remi["leaderboard_source"] = "leaderboard_remi"
df_all_raw = pd.concat([df, df_remi], ignore_index=True, sort=False)

BACKBONE_ORDER_EXT = BACKBONE_ORDER + [
    "tf_efficientnetv2_s.in21k_ft_in1k",
    "swin_tiny_patch4_window7_224",
]
BACKBONE_LABEL_EXT = {
    **BACKBONE_LABEL,
    "tf_efficientnetv2_s.in21k_ft_in1k": "EfficientNetV2-S",
    "swin_tiny_patch4_window7_224": "Swin-Tiny",
}
BACKBONE_LABEL_ORDER_EXT = [BACKBONE_LABEL_EXT[b] for b in BACKBONE_ORDER_EXT]
BACKBONE_COLORS_EXT = dict(
    zip(BACKBONE_LABEL_ORDER_EXT, sns.color_palette("colorblind", len(BACKBONE_ORDER_EXT)))
)

GRID_NAMES = {"grid_search_transfer_2", "transfer-sweep"}
df_ext = df_all_raw[
    (df_all_raw.name.isin(GRID_NAMES))
    & (df_all_raw.status == "ok")
    & (df_all_raw["data.n_images"] == 15000)
    & (df_all_raw["train.nb_epoch"] == 10)
].copy()

df_ext["group"] = df_ext.apply(assign_group, axis=1)
df_ext["backbone"] = df_ext["model.transfer.backbone"].map(BACKBONE_LABEL_EXT)
assert (df_ext["group"] != "OTHER").all(), (
    "certaines runs ne correspondent à aucun bloc de la grille"
)
df_ext["val_avg2"] = df_ext[["val_f1_macro", "val_mcc"]].mean(axis=1)

print(
    f"lignes retenues : {len(df_ext)} ({(df_ext.leaderboard_source == 'leaderboard').sum()} existantes + "
    f"{(df_ext.leaderboard_source == 'leaderboard_remi').sum()} nouvelles)"
)
df_ext.groupby(["leaderboard_source", "group"]).size().unstack(fill_value=0)
```

### 3bis. Stratégie de gel (existant + nouveau)

```python
strategy_rows_ext = []
for b in BACKBONE_ORDER_EXT:
    lbl = BACKBONE_LABEL_EXT[b]
    sub_all = df_ext[(df_ext["model.transfer.backbone"] == b) & (df_ext.group == "G1_linear_probe")]
    strategy_rows_ext.append(
        {
            "backbone": lbl,
            "strategy": "Gelé (linear probing)",
            "val_avg2": sub_all["val_avg2"].max(),
        }
    )
    for fz, label in [("1", "Partiel (1 dernier bloc)"), ("2", "Partiel (2 derniers blocs)")]:
        sub = df_ext[
            (df_ext["model.transfer.backbone"] == b) & (df_ext["model.transfer.freeze"] == fz)
        ]
        strategy_rows_ext.append(
            {"backbone": lbl, "strategy": label, "val_avg2": sub["val_avg2"].max()}
        )
    sub_none = df_ext[
        (df_ext["model.transfer.backbone"] == b) & (df_ext.group == "G2_finetune_full")
    ]
    strategy_rows_ext.append(
        {
            "backbone": lbl,
            "strategy": "Dégelé (fine-tuning complet)",
            "val_avg2": sub_none["val_avg2"].max(),
        }
    )
strat_df_ext = pd.DataFrame(strategy_rows_ext)
strategy_order = [
    "Gelé (linear probing)",
    "Partiel (1 dernier bloc)",
    "Partiel (2 derniers blocs)",
    "Dégelé (fine-tuning complet)",
]

fig, ax = plt.subplots(figsize=(13, 5.5))
sns.barplot(
    data=strat_df_ext,
    x="backbone",
    y="val_avg2",
    hue="strategy",
    order=BACKBONE_LABEL_ORDER_EXT,
    hue_order=strategy_order,
    palette="colorblind",
    ax=ax,
)
ax.set_xlabel("")
ax.set_ylabel("Meilleur val_avg2 (F1 + MCC)")
ax.set_title(
    "Stratégie de gel — meilleure configuration par backbone (existant + nouveau)",
    fontweight="bold",
)
ax.tick_params(axis="x", rotation=15)
for label in ax.get_xticklabels():
    label.set_ha("right")
ax.set_ylim(0, 1.14)
ax.legend(
    frameon=False, fontsize=8.5, ncol=4, title="", loc="upper center", bbox_to_anchor=(0.5, 1.0)
)
sns.despine(ax=ax)
fig.tight_layout()
save_fig(fig, "transfer_grid_freeze_strategy.png")
```
